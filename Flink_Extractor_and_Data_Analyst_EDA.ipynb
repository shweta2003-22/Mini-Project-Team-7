# Part 1: Data Extractor — Data Extraction & Preprocessing Pipeline

## Step 1: Environment Setup & Library Imports
Import essential data manipulation, numerical computing, visualization, and file-system libraries (`pandas`, `numpy`, `matplotlib`, `os`, `pathlib`) and configure `pandas` display options to inspect all columns and rows without truncation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 150)

print("Libraries imported successfully!")
```

## Step 2: Data Extraction & Initial Inspection
Load the raw October 2026 Flink frozen product dataset (`Flink_Frozen_Product_Data_October_2026.csv`), create a working copy (`df`), verify the dimensions (`1,134` rows × `21` columns), and preview the first 10 records alongside the full list of column names.

```python
FILE_PATH = "Flink_Frozen_Product_Data_October_2026.csv"

raw_df = pd.read_csv(FILE_PATH)

df = raw_df.copy()

print("DATA EXTRACTION COMPLETED")
print("-" * 50)
print("File:", FILE_PATH)
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

display(df.head(10))

print("\nDataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

## Step 3: Structural Data Inspection & Schema Profiling
Examine column data types, non-null counts, missing value counts, unique value cardinality, and summary statistics (`describe`) across all numerical and categorical attributes to understand the baseline structure of the dataset.

```python
df.info()

display(df.describe(include="all").T)

structure = pd.DataFrame({
    "Column_Name": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Non_Null_Count": df.notnull().sum().values,
    "Missing_Count": df.isnull().sum().values,
    "Unique_Values": df.nunique().values
})

display(structure)
```

## Step 4: Data Quality Assessment
Audit the dataset for common data engineering anomalies prior to cleaning:
1. **Missing Values:** Calculate missing counts and percentages per column.
2. **Exact Duplicates:** Identify identical rows (`14` duplicate records).
3. **Negative Quantities:** Check stock and sales volume columns for invalid negative numbers.
4. **Invalid Discounts:** Verify that `Discount_Percentage` falls strictly within `0%–100%`.
5. **Categorical Consistency:** Inspect unique values in `Category`, `Brand`, and `Warehouse_ID` for casing, whitespace, or spelling discrepancies.

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_report = missing_report.sort_values(
    by="Missing_Count",
    ascending=False
)

print("MISSING VALUE REPORT")
display(missing_report)

print("Exact duplicate rows:", df.duplicated().sum())

quantity_columns = [
    "Initial_Stock",
    "Quantity_Sold",
    "Current_Stock",
    "Expired_Quantity"
]

print("\nNEGATIVE QUANTITY CHECK")

for column in quantity_columns:
    print(
        column,
        "Negative values:",
        (df[column] < 0).sum()
    )

print(
    "\nDiscount outside 0-100:",
    (~df["Discount_Percentage"].between(0, 100)).sum()
)

print("\nCATEGORY VALUES")
print(df["Category"].value_counts(dropna=False))

print("\nBRAND VALUES")
print(df["Brand"].value_counts(dropna=False))

print("\nWAREHOUSE VALUES")
print(df["Warehouse_ID"].value_counts(dropna=False))
```

## Step 5: Duplicate Record Removal
Drop exact duplicate rows (`14` records removed, leaving `1,120` unique records) while preserving the first occurrence, and reset the DataFrame index.

```python
before_removal = len(df)

df = df.drop_duplicates(keep="first").copy()

after_removal = len(df)

removed_records = before_removal - after_removal

print("DUPLICATE REMOVAL COMPLETED")
print("Original Records:", before_removal)
print("Remaining Records:", after_removal)
print("Duplicates Removed:", removed_records)

df.reset_index(drop=True, inplace=True)
```

## Step 6: Text Standardization & Categorical Normalization
Strip leading and trailing whitespace across all text columns and map inconsistent category and brand labels (e.g., `icecream` → `Ice Cream`, `PIZZA` → `Pizza`, `OCEANWAY` → `OceanWay`) to standardized canonical names.

```python
text_columns = df.select_dtypes(
    include=["object"]
).columns

for column in text_columns:
    df[column] = df[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

category_mapping = {
    "icecream": "Ice Cream",
    "ICECREAM": "Ice Cream",
    "pizza": "Pizza",
    "PIZZA": "Pizza",
    "Sea Food": "Seafood",
    "sea food": "Seafood",
    "Frozen vegetables": "Vegetables",
    "vegetables": "Vegetables"
}

df["Category"] = df["Category"].replace(category_mapping)

brand_mapping = {
    "OCEANWAY": "OceanWay",
    "frostbite": "FrostBite",
    "Sweet Peak": "SweetPeak",
    "Green farm": "GreenFarm"
}

df["Brand"] = df["Brand"].replace(brand_mapping)

print("TEXT STANDARDIZATION COMPLETED")

print("\nCategories after cleaning:")
print(df["Category"].value_counts(dropna=False))

print("\nBrands after cleaning:")
print(df["Brand"].value_counts(dropna=False))
```

## Step 7: Data Type Conversion
Cast date fields (`Purchase_Date`, `Expiry_Date`, `Sale_Date`) to `datetime64[ns]` and numerical fields (`Purchase_Cost`, `Selling_Price`, stock quantities, discounts, and revenue) to numeric data types, coercing malformed entries to `NaN`.

```python
date_columns = [
    "Purchase_Date",
    "Expiry_Date",
    "Sale_Date"
]

numeric_columns = [
    "Purchase_Cost",
    "Selling_Price",
    "Storage_Temperature_C",
    "Initial_Stock",
    "Quantity_Sold",
    "Current_Stock",
    "Expired_Quantity",
    "Unit_Selling_Price",
    "Discount_Percentage",
    "Total_Revenue"
]

for column in date_columns:
    df[column] = pd.to_datetime(
        df[column],
        errors="coerce"
    )

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print("DATA TYPE CONVERSION COMPLETED")

print(df.dtypes)
```

## Step 8: Missing Value Flagging (Non-Destructive Handling)
Preserve unknown values as `NaN` rather than applying unsafe automatic imputation, and generate explicit boolean missing-value indicator flags (`*_Missing`) for key operational and financial columns (`Brand`, `Purchase_Cost`, `Warehouse_ID`, `Expiry_Date`, `Quantity_Sold`, `Transaction_ID`).

```python
missing_before = df.isnull().sum()

print("MISSING VALUES BEFORE HANDLING")
display(missing_before[missing_before > 0])


important_columns = [
    "Brand",
    "Purchase_Cost",
    "Warehouse_ID",
    "Expiry_Date",
    "Quantity_Sold",
    "Transaction_ID"
]

for column in important_columns:
    df[column + "_Missing"] = df[column].isnull()

print("\nMISSING VALUE FLAGS CREATED")

display(
    df[
        [c + "_Missing" for c in important_columns]
    ].head()
)
```

## Step 9: Business Rule Validation & Outlier Detection
Evaluate records against domain-specific business rules and statistical thresholds:
- Negative quantities (`Quantity_Sold < 0`)
- Out-of-bounds discounts (`Discount_Percentage` outside `0–100`)
- Non-positive costs or selling prices (`<= 0`)
- Chronological date anomalies (`Expiry_Date < Purchase_Date` or `Sale_Date > Expiry_Date`)
- Revenue calculation discrepancies (`Total_Revenue` vs. `Quantity_Sold × Unit_Selling_Price`)
- Missing transaction IDs and Interquartile Range (IQR) outliers in `Purchase_Cost`

```python
df["Invalid_Quantity"] = df["Quantity_Sold"] < 0

df["Invalid_Discount"] = (
    ~df["Discount_Percentage"].between(0, 100)
)

df["Invalid_Purchase_Cost"] = df["Purchase_Cost"] <= 0

df["Invalid_Selling_Price"] = (
    df["Selling_Price"] <= 0
)

df["Invalid_Date_Order"] = (
    df["Expiry_Date"] < df["Purchase_Date"]
)

df["Sale_After_Expiry"] = (
    df["Sale_Date"] > df["Expiry_Date"]
)

expected_revenue = (
    df["Quantity_Sold"] *
    df["Unit_Selling_Price"]
)

df["Revenue_Mismatch"] = (
    (df["Total_Revenue"] - expected_revenue).abs() > 0.01
)

df["Missing_Transaction_Flag"] = (
    df["Transaction_ID"].isnull()
)

Q1 = df["Purchase_Cost"].quantile(0.25)
Q3 = df["Purchase_Cost"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df["Purchase_Cost_Outlier"] = (
    (df["Purchase_Cost"] < lower_limit) |
    (df["Purchase_Cost"] > upper_limit)
)

validation_columns = [
    "Invalid_Quantity",
    "Invalid_Discount",
    "Invalid_Purchase_Cost",
    "Invalid_Selling_Price",
    "Invalid_Date_Order",
    "Sale_After_Expiry",
    "Revenue_Mismatch",
    "Missing_Transaction_Flag",
    "Purchase_Cost_Outlier"
]

validation_report = pd.DataFrame({
    "Issue": validation_columns,
    "Affected_Records": [
        int(df[column].sum())
        for column in validation_columns
    ]
})

print("BUSINESS RULE VALIDATION REPORT")
display(validation_report)
```

## Step 10: Data Quality Scoring & Status Assignment
Aggregate critical validation flags (`Invalid_Quantity`, `Invalid_Discount`, `Invalid_Purchase_Cost`, `Invalid_Selling_Price`, `Invalid_Date_Order`, `Revenue_Mismatch`) into `Quality_Issue_Count` and classify each row as either **`Passed`** (`1,108` records) or **`Needs Review`** (`12` records).

```python
critical_flags = [
    "Invalid_Quantity",
    "Invalid_Discount",
    "Invalid_Purchase_Cost",
    "Invalid_Selling_Price",
    "Invalid_Date_Order",
    "Revenue_Mismatch"
]

df["Quality_Issue_Count"] = (
    df[critical_flags].sum(axis=1)
)

df["Data_Quality_Status"] = np.where(
    df["Quality_Issue_Count"] > 0,
    "Needs Review",
    "Passed"
)

print("DATA QUALITY STATUS")
print(df["Data_Quality_Status"].value_counts())

display(
    df[
        ["Product_ID", "Batch_ID",
         "Transaction_ID", "Quality_Issue_Count",
         "Data_Quality_Status"]
    ].head(20)
)
```

## Step 11: Dataset Partitioning (Clean vs. Review Queues)
Split the processed DataFrame into two distinct subsets: `clean_df` (records passing all critical validation checks, ready for downstream business analysis) and `review_df` (flagged records isolated for manual investigation).

```python
clean_df = df[
    df["Data_Quality_Status"] == "Passed"
].copy()

review_df = df[
    df["Data_Quality_Status"] == "Needs Review"
].copy()

print("DATA SEPARATION COMPLETED")
print("-" * 40)

print("Total Processed Records:", len(df))
print("Passed Records:", len(clean_df))
print("Records Requiring Review:", len(review_df))

print("\nCLEAN DATA PREVIEW")
display(clean_df.head())

print("\nREVIEW DATA PREVIEW")
display(review_df.head())
```

## Step 12: Final Data Quality Audit Summary
Compile a consolidated summary table (`quality_report`) tracking record counts, removed duplicates, missing value counts, and business rule violations across the entire extraction and preprocessing pipeline.

```python
quality_report = pd.DataFrame({
    "Metric": [
        "Original Records",
        "Original Columns",
        "Exact Duplicates Removed",
        "Processed Records",
        "Clean Records",
        "Records Requiring Review",
        "Missing Brand",
        "Missing Purchase Cost",
        "Missing Warehouse",
        "Missing Expiry Date",
        "Missing Quantity Sold",
        "Missing Transaction ID",
        "Negative Quantities",
        "Invalid Discounts",
        "Revenue Mismatches",
        "Sales After Expiry"
    ],

    "Value": [
        len(raw_df),
        len(raw_df.columns),
        len(raw_df) - len(raw_df.drop_duplicates()),
        len(df),
        len(clean_df),
        len(review_df),
        int(df["Brand"].isnull().sum()),
        int(df["Purchase_Cost"].isnull().sum()),
        int(df["Warehouse_ID"].isnull().sum()),
        int(df["Expiry_Date"].isnull().sum()),
        int(df["Quantity_Sold"].isnull().sum()),
        int(df["Transaction_ID"].isnull().sum()),
        int(df["Invalid_Quantity"].sum()),
        int(df["Invalid_Discount"].sum()),
        int(df["Revenue_Mismatch"].sum()),
        int(df["Sale_After_Expiry"].sum())
    ]
})

print("FINAL DATA QUALITY REPORT")
display(quality_report)
```

## Step 13: Export Processed Artifacts
Create the `Flink_Preprocessing_Output` directory and export four standardized CSV deliverables:
1. `Flink_Processed_Full_Data.csv` — Complete processed dataset with all quality flags.
2. `Flink_Clean_Data.csv` — Validated records cleared for analytical modeling.
3. `Flink_Review_Required.csv` — Anomalous records isolated for data stewardship review.
4. `Flink_Data_Quality_Report.csv` — Executive summary of pipeline quality metrics.

```python
OUTPUT_FOLDER = Path("Flink_Preprocessing_Output")

OUTPUT_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)

df.to_csv(
    OUTPUT_FOLDER / "Flink_Processed_Full_Data.csv",
    index=False
)

clean_df.to_csv(
    OUTPUT_FOLDER / "Flink_Clean_Data.csv",
    index=False
)

review_df.to_csv(
    OUTPUT_FOLDER / "Flink_Review_Required.csv",
    index=False
)

quality_report.to_csv(
    OUTPUT_FOLDER / "Flink_Data_Quality_Report.csv",
    index=False
)

print("ALL FILES EXPORTED SUCCESSFULLY")
print("\nOutput Files:")

for file in OUTPUT_FOLDER.iterdir():
    print(file.name)
```

## Step 14: Post-Export Integrity Verification
Reload `Flink_Processed_Full_Data.csv` from disk and assert exact row count (`1,120`) and column schema (`38` columns) parity with the in-memory DataFrame to confirm lossless export.

```python
final_file = (
    OUTPUT_FOLDER / "Flink_Processed_Full_Data.csv"
)

verified_df = pd.read_csv(final_file)

print("FINAL VERIFICATION")
print("-" * 40)

print("Original Dataset:", raw_df.shape)
print("Processed Dataset:", df.shape)
print("Exported Dataset:", verified_df.shape)

assert len(verified_df) == len(df), (
    "Record count mismatch!"
)

assert list(verified_df.columns) == list(df.columns), (
    "Column mismatch!"
)

print("\nSUCCESS: DATA EXTRACTION AND EXPORT VERIFIED!")
```
